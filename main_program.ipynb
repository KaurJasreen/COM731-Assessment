```python
import task_a
import task_b
import task_c

def main():
    print("=== COM731 Credit Card Customer Analysis ===")
    file_path = input("Enter the path to your CSV file (e.g. credit_card_customers.csv): ").strip()

    # Load data for Task A
    raw_data = task_a.load_csv_data(file_path)
    if not raw_data:
        return

    # Load data for Task B & C
    df = task_b.load_dataframe(file_path)
    if df is None:
        return

    while True:
        print("\n--- Main Menu ---")
        print("1. Task A - CSV Reader Analysis")
        print("2. Task B - Pandas Data Analysis")
        print("3. Task C - Data Visualization")
        print("4. Exit")

        choice = input("Choose an option (1-4): ").strip()

        if choice == '1':
            while True:
                print("\n-- Task A Menu ---")
                print("a1. Get Client Info by Card Type")
                print("a2. Get Customers by State")
                print("a3. Customers by Marital Status & Dependents")
                print("a4. Custom Filter: House Owner = No & Credit Limit > 5000")
                print("b. Back to Main Menu")
                sub = input("Choose subtask (a1-a4, b): ").strip().lower()

                if sub == 'a1':
                    task_a.task_a1(raw_data)
                elif sub == 'a2':
                    task_a.task_a2(raw_data)
                elif sub == 'a3':
                    task_a.task_a3(raw_data)
                elif sub == 'a4':
                    task_a.task_a4(raw_data)
                elif sub == 'b':
                    break
                else:
                    print("Invalid input. Please enter a1 to a4 or b only!!.")

        elif choice == '2':
            while True:
                print("\n-- Task B Menu ---")
                print("b1. Top 3 Occupations with Specific Card and High Income")
                print("b2. Avg Fee and Income by Occupation & Card")
                print("b3. Avg Fee by Marital Status (filtered)")
                print("b4. Custom Insight: Avg Interest by Education")
                print("b. Back to Main Menu")
                sub = input("Choose subtask (b1-b4, b): ").strip().lower()

                if sub == 'b1':
                    task_b.task_b1(df)
                elif sub == 'b2':
                    task_b.task_b2(df)
                elif sub == 'b3':
                    task_b.task_b3(df)
                elif sub == 'b4':
                    task_b.task_b4(df)
                elif sub == 'b':
                    break
                else:
                    print("Invalid input. Please try again.")

        elif choice == '3':
            while True:
                print("\n-- Task C Menu ---")
                print("c1. Pie Chart: Card Type by Education")
                print("c2. Line Chart: Weekly Avg Interest per Card")
                print("c3. Bar Chart: Annual Fee vs Acquisition Cost")
                print("c4. Custom Chart: Avg Transaction Amount by Education")
                print("b. Back to Main Menu")
                sub = input("Choose subtask (c1-c4, b): ").strip().lower()

                if sub == 'c1':
                    task_c.task_c1(df)
                elif sub == 'c2':
                    task_c.task_c2(df)
                elif sub == 'c3':
                    task_c.task_c3(df)
                elif sub == 'c4':
                    task_c.task_c4(df)
                elif sub == 'b':
                    break
                else:
                    print("Invalid input. Please try again.")

        elif choice == '4':
            print("Program is finished....Have a nice day!")
            break
        else:
            print("Invalid input. Please enter 1 to 4.")

if __name__ == "__main__":
    main()
```
